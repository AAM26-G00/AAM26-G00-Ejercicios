```python
import numpy as np
import pandas as pd

np.random.seed(42)  # Para poder reproducir la ejecución siempre igual

nombres_base = ['Espada de Fuego', 'Porra Legendaria', 'Pedro', 'Ana', 'Luis', 'Carmen', 'Espada de Hielo', 'Lucía', 'Miguel', 'Sofía', 'Carlos', 'Laura', 'Diego', 'María', 'Andrés', 'Elena', 'Pablo', 'Isabel', 'Javier', 'Cristina']
tipos = np.random.choice(['arma', 'armadura', 'pocion'], size=20, p=[0.5, 0.3, 0.2])
rarezas = np.random.choice(['común', 'raro', 'legendario'], size=20, p=[0.4, 0.35, 0.25])
precios = np.random.randint(50, 1500, size=20)
niveles = np.random.randint(1, 30, size=20)

# Parche a los datos sintéticos
tipos[0], rarezas[0], precios[0], niveles[0] = 'arma', 'legendario', 850, 12
tipos[6], rarezas[6], precios[6], niveles[6] = 'arma', 'legendario', 1200, 15

df_items = pd.DataFrame({
    'nombre': nombres_base,
    'tipo': tipos,
    'rareza': rarezas,
    'precio_oro': precios,
    'nivel_minimo': niveles
})

print("DataFrame de items:")
print(df_items.head())

filtro_armas_legendarias = (
    (df_items['tipo'] == 'arma') & 
    (df_items['rareza'] == 'legendario') & 
    (df_items['nivel_minimo'] <= 15) & 
    (df_items['precio_oro'] > 500)
)

df_filtrado = df_items[filtro_armas_legendarias]

print("\nItems filtrados (armas legendarias con nivel mínimo <= 15 y precio > 500):")
print(df_filtrado)
```
