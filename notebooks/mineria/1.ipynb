```python
import numpy as np

NUM_ENEMIGOS = 100

x = np.random.uniform(-50, 50, size=NUM_ENEMIGOS)
y = np.random.uniform(-50, 50, size=NUM_ENEMIGOS)

z = np.zeros(NUM_ENEMIGOS)

posiciones = np.column_stack((x, y, z))

distancias = np.linalg.norm(posiciones, axis=1)

RADIO_DETECCION = 15.0
mascaras = distancias < RADIO_DETECCION

indice_cercanos = np.where(mascaras)[0]

print(f"Dimensiones de la matriz de posiciones: {posiciones.shape}")
print(f"Dimensiones de las distancias: {distancias.shape}")
print(f"Número de enemigos detectados: {len(indice_cercanos)}")
print(f"Índices de enemigos cercanos: {indice_cercanos}")

posiciones
```
